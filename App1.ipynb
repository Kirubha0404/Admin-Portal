```python
from textblob import TextBlob

# Get text from the user
text = input("Enter a sentence: ")

# Analyze the text
analysis = TextBlob(text)

# Get polarity
polarity = analysis.sentiment.polarity

# Determine sentiment
if polarity > 0:
    sentiment = "Positive"
elif polarity < 0:
    sentiment = "Negative"
else:
    sentiment = "Neutral"

# Display results
print("\nText:", text)
print("Polarity:", polarity)
print("Sentiment:", sentiment)
```
